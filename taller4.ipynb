```python
top = df.sort_values("total_paid_usd", ascending = False).head(10)
top[["country", "platform", "product_category", "total_paid_usd"]]
```

10. Al ordenar el dataset de forma descendente nos damos cuenta que todos los productos pertenecen a la categoria de "Electronics". Lo que nos puede indicar que este tipo de producto es el mas caro dentro del catalogo.

```python
pf = (df.groupby("platform")
      .agg( promedio_pagado_plataforma = ("total_paid_usd","mean"))
      .sort_values("promedio_pagado_plataforma", ascending=False))
pf
```
